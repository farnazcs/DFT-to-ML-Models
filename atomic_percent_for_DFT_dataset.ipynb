```python

# Function to parse a formula and calculate atomic fractions
def parse_formula(formula):
    # Match elements and their counts in the formula
    matches = re.findall(r'([A-Z][a-z]*)(\d*)', formula)
    total_atoms = 0
    atom_counts = {}

    for elem, count in matches:
        count = int(count) if count else 1
        total_atoms += count
        atom_counts[elem] = atom_counts.get(elem, 0) + count

    # Calculate fractions
    fractions = {elem: atom_counts.get(elem, 0) / total_atoms for elem in elements}
    return fractions

```

```python
# Process each compound and compute atomic fractions
parsed_data = []
for compound in compounds:
    fractions = parse_formula(compound)
    parsed_data.append(fractions)

# Create output DataFrame
output_df = pd.DataFrame(parsed_data)

# Save to CSV file
output_df.to_csv(output_file, index=False)

# Print the head of the output file
print(output_df.head())
```

```python
import pandas as pd
import re

# Read the input CSV file
input_file = 'atom_percent.csv'  # Replace this with your actual file path
output_file = 'atom_per_labeled.csv'

# Read the CSV file
data = pd.read_csv(input_file)

# Extract required columns
compounds = data['Compounds']  # 'compounds' column  <-- Added
energy_per_atom = data['Energy per atom']  # 'Energy per atom' column  <-- Added

# List of elements to include in the output table
elements = ['Al', 'Mg', 'Si', 'Zn', 'Li', 'Ti', 'Ni', 'Cu', 'Ag', 'As', 'Au', 'B', 'Bi', 'C',
            'Ca', 'Cd', 'Co', 'Ga', 'Hf', 'In', 'Mo', 'Nb', 'O', 'Pb', 'P', 'S', 'Sn', 'Th',
            'V', 'W', 'Zr', 'Fe', 'Mn', 'Cr']

# Function to parse a formula and calculate atomic fractions
def parse_formula(formula):
    # Match elements and their counts in the formula
    matches = re.findall(r'([A-Z][a-z]*)(\d*)', formula)
    total_atoms = 0
    atom_counts = {}

    for elem, count in matches:
        count = int(count) if count else 1
        total_atoms += count
        atom_counts[elem] = atom_counts.get(elem, 0) + count

    # Calculate fractions
    fractions = {elem: atom_counts.get(elem, 0) / total_atoms for elem in elements}
    return fractions

# Process each compound and compute atomic fractions
parsed_data = []
for compound in compounds:
    fractions = parse_formula(compound)
    parsed_data.append(fractions)

# Create output DataFrame
parsed_df = pd.DataFrame(parsed_data)  # <-- Added

# Combine columns into final output
output_df = pd.concat([data['Compounds'], data['Energy per atom'], parsed_df], axis=1)  # <-- Added/Modified

# Save to CSV file
output_df.to_csv(output_file, index=False)

# Print the head of the output file
print(output_df.head())  # <-- Modified
```
